```python
import numpy as np
from multipie.util.util import str_to_sympy
from multipie import Group
from data_root_cluster import mapping_d6h_oh
from creation.data_parent_pg import data_representative_parent_pg
from creation.data_parent_sg import data_representative_parent_sg

from qtdraw import QtDraw
qtdraw = QtDraw()

c_to_p = data_representative_parent_pg | data_representative_parent_sg
```

```python
# plot root cluster and correspondence of point subgroup.
c_group = 152 # "D4"
p_group = c_to_p[c_group]

cg = Group(c_group)
pg = Group(p_group)

cwp = list(cg.wyckoff["site"].keys())[-1]
pwp = list(pg.wyckoff["site"].keys())[-1]
cpt = cg.wyckoff["site"][cwp]["reference"]
ppt = pg.wyckoff["site"][pwp]["reference"]

cbwp = list(cg.wyckoff["bond"].keys())[-1]
pbwp = list(pg.wyckoff["bond"].keys())[-1]
cbd = cg.wyckoff["bond"][cbwp]["reference"]
pbd = pg.wyckoff["bond"][pbwp]["reference"]
cvec, cctr = cbd[:,0:3], cbd[:,3:6]
pvec, pctr = pbd[:,0:3], pbd[:,3:6]

qtdraw.clear_data()
if type(c_group)==int:
    qtdraw.set_cell("single")
    qtdraw.set_axis("on")
else:
    qtdraw.set_cell("off")
    qtdraw.set_axis("full")
qtdraw.add_text2d(f"{str(pg)} => {str(cg)}")
qtdraw.set_crystal(cg.info.crystal)

for i in ppt:
    qtdraw.add_site(position=i, size=0.02, color="silver", name=f"parent")
for i in cpt:
    qtdraw.add_site(position=i, size=0.04, color="salmon", name=f"child", opacity=0.5)

for i, v in zip(pctr, pvec):
    qtdraw.add_vector(direction=v, position=i, length=-0.3, color="silver")
for i, v in zip(cctr, cvec):
    qtdraw.add_vector(direction=v, position=i, length=-0.3, color="salmon", opacity=0.5)
```
