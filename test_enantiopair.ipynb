```python
from multipiegen import __top_dir__
import os
os.chdir(__top_dir__)

from multipie import Group
from multipiegen.data.data_parent_sg import data_parent_sg
```

```python
enantio_pair = [(76,78),(91,95),(92,96),(144,145),(151,153),(152,154),(169,170),(171,172),(178,179),(180,181),(212,213)]

print("=== site ===")
for right, left in enantio_pair:
    gr = Group(right)
    gl = Group(left)
    gr_p = data_parent_sg[right]
    gl_p = data_parent_sg[left]
    assert gr_p.keys()==gl_p.keys() # same parent.
    no_p = list(gr_p.keys())[0]
    g = Group(no_p)
    r_mp, r_R = gr_p[no_p]
    l_mp, l_R = gl_p[no_p]
    assert r_mp==l_mp # same mapping.
    assert r_R==l_R # same conversion matrix.
    print(f"({gr.info.international},{gl.info.international}) => {g.info.international}")
    for wp_r, wp_l in zip(gr.wyckoff["site"].keys(),gl.wyckoff["site"].keys()):
        assert wp_r==wp_l # same wyckoff.
        if gr.wyckoff["site"][wp_r]["mapping"] != gl.wyckoff["site"][wp_l]["mapping"]:
            print("R and L op-mapping differ at", wp_r)
            print("  R:", gr.wyckoff["site"][wp_r]["mapping"])
            print("  L:", gl.wyckoff["site"][wp_l]["mapping"])
print("=== bond ===")
for right, left in enantio_pair:
    gr = Group(right)
    gl = Group(left)
    gr_p = data_parent_sg[right]
    gl_p = data_parent_sg[left]
    assert gr_p.keys()==gl_p.keys() # same parent.
    no_p = list(gr_p.keys())[0]
    g = Group(no_p)
    r_mp, r_R = gr_p[no_p]
    l_mp, l_R = gl_p[no_p]
    assert r_mp==l_mp # same mapping.
    assert r_R==l_R # same conversion matrix.
    print(f"({gr.info.international},{gl.info.international}) => {g.info.international}")
    for wp_r, wp_l in zip(gr.wyckoff["bond"].keys(),gl.wyckoff["bond"].keys()):
        assert wp_r==wp_l # same wyckoff.
        if gr.wyckoff["bond"][wp_r]["mapping"] != gl.wyckoff["bond"][wp_l]["mapping"]:
            print("R and L mapping differ at", wp_r)
            print("  R:", gr.wyckoff["bond"][wp_r]["mapping"])
            print("  L:", gl.wyckoff["bond"][wp_l]["mapping"])
```
