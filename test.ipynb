```python
def shift(a, var):
    a = np.vectorize(lambda e: sp.expand(e - e.subs({var[0]: 0, var[1]: 0, var[2]: 0})) + e.subs({var[0]: 0, var[1]: 0, var[2]: 0}) % 1, otypes=[object])(a)
    return a

def inversion(a, var):
    a = np.vectorize(lambda i: -i.subs({"x":-var[0],"y":-var[1],"z":-var[2]}))(a)
    return shift(a, var)

var = Group.global_info()["harmonics"]["variable"]
c_group = 198
cg = Group(c_group)
wp = cg.wyckoff["site"]["4a"]["conventional"]
wpi = inversion(wp, var)
str(wp.tolist()).replace(" ","")==str(wpi.tolist()).replace(" ","")
```
