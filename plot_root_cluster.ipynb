```python
from multipiegen import __top_dir__
import os
os.chdir(__top_dir__)

import numpy as np
from multipie.util.util import str_to_sympy
from multipiegen.data.data_root_cluster import root_cluster_cart, root_cluster_cart_site0, oh_cluster_frac, d6h_cluster_frac
from multipiegen.data.data_op_mapping import op_mapping_root_cluster
from qtdraw import QtDraw
qtdraw = QtDraw()

p = str_to_sympy(root_cluster_cart_site0).astype(float)
p = p / np.sqrt(p@p)

qtdraw.set_cell("off")
qtdraw.set_crystal("cubic")
qtdraw.set_view([1,1,1])
```

```python
# plot root cluster from mapping of Oh and D6h.
cluster = str_to_sympy(root_cluster_cart, subs={"x": p[0], "y": p[1], "z": p[2]}).astype(float)
qtdraw.clear_data()
qtdraw.set_crystal("cubic")
qtdraw.add_site(size=1,color="snow",opacity=0.1)
colors = ["moss","salmon","cantaloupe","aqua"]
for group in ["Oh", "D6h"]:
    for n, mp in enumerate(np.array(op_mapping_root_cluster[group][1])):
        color = colors[n] if group=="D6h" else "silver"
        for no, i in enumerate(cluster[mp-1]):
            qtdraw.add_site(position=i, size=0.04, color=color, name=f"{group}{n}", label=f"p{no+1}")
```

```python
# plot root cluster and categorize Oh and D6h subsets.
cluster = str_to_sympy(root_cluster_cart, subs={"x": p[0], "y": p[1], "z": p[2]}).astype(float)
qtdraw.clear_data()
qtdraw.set_crystal("cubic")
qtdraw.add_site(size=1,color="snow",opacity=0.1)
for no, i in enumerate(cluster):
    for n1, mp1 in enumerate(op_mapping_root_cluster["D6h"][1]):
        for n2, mp2 in enumerate(op_mapping_root_cluster["Oh"][1]):
            c = "salmon" if n2==0 else "aqua"
            if no+1 in mp1 and no+1 in mp2:
                qtdraw.add_site(position=i, size=0.04, color=c, name=f"d6h{n1}_oh{n2}", label=f"p{no+1}")
```

```python
# plot root cluster and correspondence of point subgroup.
group = "D3"

cluster = str_to_sympy(root_cluster_cart, subs={"x": p[0], "y": p[1], "z": p[2]}).astype(float)
mapping = op_mapping_root_cluster[group][1]
qtdraw.clear_data()
qtdraw.set_crystal("cubic")
qtdraw.add_site(size=1,color="snow",opacity=0.1)
for no, i in enumerate(cluster):
    for nm, mp in enumerate(mapping):
        color = "silver" if nm!=0 else "salmon"
        if no+1 in mp:
            qtdraw.add_site(position=i, size=0.04, color=color, name=f"set{mp}", label=f"p{no+1}")
```

```python
# plot oh cluster.
cluster = str_to_sympy(oh_cluster_frac, subs={"x": p[0], "y": p[1], "z": p[2]}).astype(float)

qtdraw.clear_data()
qtdraw.set_crystal("cubic")
qtdraw.add_site(size=1,color="snow",opacity=0.1)
for no, i in enumerate(cluster):
    qtdraw.add_site(position=i, size=0.04, color="silver", name=f"oh", label=f"p{no+1}")
```

```python
# plot d6h cluster.
cluster = str_to_sympy(d6h_cluster_frac, subs={"x": p[0], "y": p[1], "z": p[2]}).astype(float)

qtdraw.clear_data()
qtdraw.set_crystal("hexagonal")
qtdraw.add_site(size=1,color="snow",opacity=0.1)
for no, i in enumerate(cluster):
    qtdraw.add_site(position=i, size=0.04, color="silver", name=f"d6h", label=f"p{no+1}")
```
