```python
import numpy as np
from multipie.util.util import str_to_sympy
from data_root_cluster import root_cluster, mapping_root_cluster
from qtdraw import QtDraw
qtdraw = QtDraw()

p = np.array([0.1, 0.23, 0.31])
p = p / np.sqrt(p@p)
cluster = str_to_sympy(root_cluster, subs={"x": p[0], "y": p[1], "z": p[2]}).astype(float)

qtdraw.set_cell("off")
qtdraw.set_view([1,1,1])
```

```python
qtdraw.clear_data()
qtdraw.add_site(size=1,color="snow",opacity=0.1)
for group in ["Oh", "D6h"]:
    for n, mp in enumerate(np.array(mapping_root_cluster[group])):
        for no, i in enumerate(cluster[mp-1]):
            qtdraw.add_site(position=i, size=0.04, color="silver", name=f"{group}{n}", label=f"p{no+1}")
```

```python
qtdraw.clear_data()
qtdraw.add_site(size=1,color="snow",opacity=0.1)
for no, i in enumerate(cluster):
    for n1, mp1 in enumerate(mapping_root_cluster["D6h"]):
        for n2, mp2 in enumerate(mapping_root_cluster["Oh"]):
            c = "salmon" if n2==0 else "aqua"
            if no+1 in mp1 and no+1 in mp2:
                qtdraw.add_site(position=i, size=0.04, color=c, name=f"d6h{n1}_oh{n2}", label=f"p{no+1}")
```

```python
group = "D3"
mapping = mapping_root_cluster[group]
qtdraw.clear_data()
qtdraw.add_site(size=1,color="snow",opacity=0.1)
for no, i in enumerate(cluster):
    for mp in mapping:
        if no+1 in mp:
            qtdraw.add_site(position=i, size=0.04, color="silver", name=f"set{mp}", label=f"p{no+1}")
```
