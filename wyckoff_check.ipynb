```python
import numpy as np
import sympy as sp
from multipie import Group
# from qtdraw import QtDraw
from data_parent_sg import data_parent_sg
```

```python
def check_each_sg(c_group):
    dic = data_parent_sg[c_group]
    p_group = list(dic.keys())[0]
    print(f"{p_group} => {c_group}")
    cg = Group(c_group)
    pg = Group(p_group)
    mapping = np.array(data_parent_sg[c_group][p_group][0][0])-1
    pg_wp = list(pg.wyckoff["site"].keys())[-1]
    cg_wp = list(cg.wyckoff["site"].keys())[-1]

    pw = pg.wyckoff["site"][pg_wp]["conventional"][mapping]
    cw = cg.wyckoff["site"][cg_wp]["conventional"]
    site_err = not bool(np.all(pw==cw))
    if site_err:
        print(mapping)
        print(pw.tolist())
        print(cw.tolist())

    pw = pg.wyckoff["bond"][pg.wyckoff["site"][pg_wp]["bond"][0]]["conventional"][mapping]
    cw = cg.wyckoff["bond"][cg.wyckoff["site"][cg_wp]["bond"][0]]["conventional"]
    bond_err = not bool(np.all(pw==cw))
    if bond_err:
        print(mapping)
        print(pw.tolist())
        print(cw.tolist())

    return site_err, bond_err

def check_sg():
    cnt_site = 0
    cnt_bond = 0
    for c_group in data_parent_sg.keys():
        site_err, bond_err = check_each_sg(c_group)
        if site_err:
            cnt_site += 1
        if bond_err:
            bond_err += 1
    if cnt_site>0:
        print(f"found {cnt_site} error in site wyckoff.")
    if cnt_bond>0:
        print(f"found {cnt_bond} error in bond wyckoff.")
    if cnt_site==0 and cnt_bond==0:
        print("no error.")
check_sg()
```
