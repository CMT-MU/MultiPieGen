```python
from multipiegen import __top_dir__
import os
os.chdir(__top_dir__)

import numpy as np
import sympy as sp
from multipie import Group
# from qtdraw import QtDraw
from util_spherical_harmonics import Olm
```

```python
slst = {}
for l in range(12):
    for m in range(l, 0,-1):
        c = sp.simplify((sp.S(-1)**m*Olm(l,m)+Olm(l,-m))/sp.sqrt(2))
        s = sp.simplify((sp.S(-1)**m*Olm(l,m)-Olm(l,-m))/sp.sqrt(2)/sp.I)
        slst[f"C{l},{m}"] = str(c).replace(" ","").replace("*","")
        slst[f"S{l},{m}"] = str(s).replace(" ","").replace("*","")
    slst[f"C{l},0"] = str(sp.simplify(Olm(l,0))).replace(" ","").replace("*","")
slst
```
