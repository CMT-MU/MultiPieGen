```python
from multipiegen import __top_dir__
import os
os.chdir(__top_dir__)

import numpy as np
from multipie.util.util import str_to_sympy
from multipie import Group
from multipiegen.data.data_op_mapping import op_mapping_d6h_oh_cluster
from multipiegen.data.data_parent_pg import data_representative_parent_pg
from multipiegen.data.data_parent_sg import data_representative_parent_sg

from qtdraw import QtDraw
qtdraw = QtDraw()

c_to_p = data_representative_parent_pg | data_representative_parent_sg
```

```python
# plot root cluster and correspondence of point subgroup.
c_group = 198
p_group = c_to_p[c_group]

cg = Group(c_group)
pg = Group(p_group)

if "plus_set" in pg.symmetry_operation.keys():
    p_plus_set = pg.symmetry_operation["plus_set"].astype(float)
else:
    p_plus_set = np.array([[0,0,0]], dtype=float)
if "plus_set" in cg.symmetry_operation.keys():
    c_plus_set = cg.symmetry_operation["plus_set"].astype(float)
else:
    c_plus_set = np.array([[0,0,0]], dtype=float)

cwp = list(cg.wyckoff["site"].keys())[-1]
pwp = list(pg.wyckoff["site"].keys())[-1]
cpt = cg.wyckoff["site"][cwp]["reference"].astype(float)
ppt = pg.wyckoff["site"][pwp]["reference"].astype(float)

cbwp = list(cg.wyckoff["bond"].keys())[-1]
pbwp = list(pg.wyckoff["bond"].keys())[-1]
cbd = cg.wyckoff["bond"][cbwp]["reference"].astype(float)
pbd = pg.wyckoff["bond"][pbwp]["reference"].astype(float)
cvec, cctr = cbd[:,0:3], cbd[:,3:6]
pvec, pctr = pbd[:,0:3], pbd[:,3:6]

qtdraw.clear_data()
if type(c_group)==int:
    qtdraw.set_cell("single")
    qtdraw.set_axis("on")
else:
    qtdraw.set_cell("off")
    qtdraw.set_axis("full")
qtdraw.add_text2d(f"{str(pg)} => {str(cg)}")
qtdraw.set_crystal(cg.info.crystal)

for i in ppt:
    for t in p_plus_set:
        qtdraw.add_site(position=i+t, size=0.02, color="silver", name=f"parent")
for i in cpt:
    for t in c_plus_set:
        qtdraw.add_site(position=i+t, size=0.04, color="salmon", name=f"child", opacity=0.5)

for i, v in zip(pctr, pvec):
    for t in p_plus_set:
        qtdraw.add_vector(direction=v, position=i+t, length=-0.3, color="silver", name=f"parent")
for i, v in zip(cctr, cvec):
    for t in c_plus_set:
        qtdraw.add_vector(direction=v, position=i+t, length=-0.3, color="salmon", opacity=0.5, name=f"child")
```
