```python
from multipiegen import __top_dir__
import os
os.chdir(__top_dir__)

import numpy as np
import sympy as sp
from multipie import Group
from multipie.util.util import str_to_sympy
from qtdraw import QtDraw
qtdraw = QtDraw()
```

```python
from multipiegen.data.data_root_cluster import oh_cluster_cart, d6h_cluster_cart, oh_cluster_cart_site0, d6h_cluster_cart_site0, oh_cluster_cart_vector0, d6h_cluster_cart_vector0

site0 = str_to_sympy(oh_cluster_cart_site0)
site0h = str_to_sympy(d6h_cluster_cart_site0)
vector0 = str_to_sympy(oh_cluster_cart_vector0)
vector0h = str_to_sympy(d6h_cluster_cart_vector0)
```

```python
oh_pts = str_to_sympy(oh_cluster_cart, subs={"x":site0[0], "y": site0[1], "z": site0[2]})
d6h_pts = str_to_sympy(d6h_cluster_cart, subs={"x":site0h[0], "y": site0h[1], "z": site0h[2]})
```

```python
qtdraw.clear_data()
for p in oh_pts:
    qtdraw.add_site(position=p, size=0.2, color="salmon", name="oh")
for p in d6h_pts:
    qtdraw.add_site(position=p, size=0.2, color="aqua", name="d6h")
```
